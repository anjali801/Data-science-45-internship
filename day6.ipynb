```python

class address:
  def __init__(self,city,state):
    self.city = city
    self.state = state

  def location(self):
   return f"my location is  {self.city} in {self.state}"

class user(address):
  def __init__(self,name,age,city,state):
    self.name = name
    self.age = age
    self.city = city
    self.state = state

  def userName(self):
    print(f" My Name is {self.name}")

  def userDetail(self):
    print(f"My Name is {self.name} and  My Location is {self.city} in {self.state}")


u = user("Anjali",20 , "Jaipur","Rajasthan")
u.userName()
u.userDetail()
u.location()                                      # here is the inheritance
print(  u.location())
```

```python
# Example of super keyword :-
class address:
  def __init__(self,city,state):
    self.city = city
    self.state = state

  def location(self):
   return f"My location is  {self.city} in {self.state}"

class user(address):
  def __init__(self,name,age,city,state):
    super().__init__(city,state)                            # here is use of super keyword
    self.name = name
    self.age = age

  def userName(self):
    print(f"My Name is {self.name}")

  def userDetail(self):
    print(f"My Name is {self.name} and  My Location is {self.city} in {self.state}")


u = user("Anjali",20 , "Jaipur","Rajasthan")
print(u.city)
u.userName()
u.userDetail()
print(  u.location())                                    # here is the inheritance
```

```python
# Encapsulation:-
class address:
  def __init__(self,city,state):
    self.__city = city                                            # Here self.__city become private by the use of  __
    self.state = state

  def location(self):
    print(f"My location is  {self.__city} in {self.state}")

  def getCity(self):
     return self.__city

a = address("chhapra","Bihar")
a.location()
# print(a.city) _ >
print(a.getCity())
print(a.state)
```

```python
# polymorphism :-

class address:
  def __init__(self,city, state):
    self.city = city
    self.state = state
  def location(self):
    print(f"My location is  {self.city} in {self.state}")

class Hometown:
  def __init__(self,city,state):
    self.city = city
    self.state = state
  def location(self):
    print(f"My Hometown is  {self.city} in {self.state}")

a = address("vadoda","Gujrat")
a.location()

b = Hometown("jaipur","Rajasthan")
b.location()
```
