# feature scaling :   

Feature scaling is a data preprocessing technique used in machine learning to normalize the range of independent variables (features). Because features often have vastly different scales (e.g., age from 0-100 vs. income from $0-$1,000,000), scaling ensures that all variables contribute equally to model performance and stability

```python
# Data processing :


# -> Handle issing values :---
import pandas as pd
data = {
    "Name" : ["Ram" ,"Kamal","Ajay"],
    "Age" : [25,None,32],
    "salary" : [5000,6000,None]
}
df = pd.DataFrame(data)
print("original Dataframe")
print(df)



#null findout | fillna | dropna
print(df.isnull().sum())


df_drop = df.dropna()
print(df_drop)

#fillna
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['salary'].fillna(df['salary'].mean(),inplace=True)
print(df)
```
