```python
# Nested dictionary
d = {"address":{"address1":{"city":"kukas city","district":"jaipur"},
                "address2":{"city":"gopalpura","district":"arya main"}}}
print(d)
print(type(d))
print(d["address"])
print(d["address"]["address1"])
print(d["address"]["address2"])
print(d["address"]["address1"]["city"])
print(d["address"]["address2"]["city"])
print(d["address"]["address1"]["district"])
print(d["address"]["address2"]["district"])
print("keys of the address1 : ",d["address"]["address1"].keys())
print("keys of the address2 : ",d["address"]["address2"].keys())
print("values of the address1 : ",d["address"]["address1"].values())
print("values of the address2 : ",d["address"]["address2"].values())
```

```python
# Nested List
l = [10,20,30,["hello","hello1","hello2"],[True,False]]
print(l)
print(l[0])
print(l[1])
print(l[2])
print(l[3])
for i in l[3]:
  print(i)
print(l[4])
for i in l[4]:
  print(i)


#slicing
print("\n example of slicing:")
l1 =l[:3]
print(l1)
l2 = l[2:4]
print(l2)
l3 = l[2:]
print(l3)
```

```python
# access kara hai all the element :
l = [10,20,30,["hello","hello1","hello2",[True,False]]]
print(l[3])
for i in l[3]:
  print(i)
for i in l[3][3]:
  print(i)

l2 = l[3:]
print(l2)

l3 = l[3][1:3]
print(l3)

```
