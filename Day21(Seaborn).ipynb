# Pairplot :               
seaborn.pairplot()

->is used for visualizing relationships between multiple variables in a dataset. By creating a grid of scatter plots it helps to identify how different features interact with each other to identify patterns, correlations and trends in data.

-> generally means all the possible plots are made of the datasets through the pairplot



```python
sns.pairplot(var1)
```

```python
sns.set_style('dark')
sns.pairplot(var1)
```

```python
sns.pairplot(var1,hue="sex")
```

```python
sns.pairplot(var1, vars =["total_bill","tip"],hue="sex")           # vars = used to pass the particular parameter
```

```python
sns.pairplot(var1, vars =["total_bill","tip"],hue="sex",kind = "reg")
```

```python
sns.pairplot(var1, vars =["total_bill","tip"],hue="sex", diag_kind= " hist")
```

# Violin Plot :                        

A violin plot plays a similar role as a box and whisker plot .

It shows the distribution of quantitative  dat across several levels of one ( or more ) categorical variables such that those distributions can be compared
