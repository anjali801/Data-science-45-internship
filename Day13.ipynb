```python
import pandas as pd
data = {
    "Emp_ID" : [101,102,103,104,105],
    "Name": ["Anna","Jiva","NIkhil","Elsa","Ri"],
    "Department":["IT","fashion","Music","cse","game"],
    "Salary" : [50000,4000000,67000,23000,34000],
    "Experiencce":[1,4,5,3,2]
}

df = pd.DataFrame(data)
print(df)

print("\n",df.head(3))
print("\n",df.tail(2))
print("\n",df.head(-2))
print("\n",df.tail(-3))
print("\n",df.shape)
print("\n",df.info())
print("\n",df.info(verbose=False))
print("\n",df.describe())
df.rename(columns={"deparment":"Field"})
```
