```python
import numpy as np                             # same as * (multiple) , ** (power) , / (divide) ,% (mod)
data = [1,2,3,4,5]
data1 = [6,7,8,9,10]
x = np.array(data)
y = np.array(data1)
x1 = x + x
y1 = y-y
x2 = np.add(x,x)
y2 = np.subtract(y,y)
print(x1)
print(y1)
print(x2)
print(y2)
```

```python
# mean and mediam
import numpy as np
a = np.array([1,2,3,4,5,6 ,7 ,10])
print(np.mean(a))
print(np.median(a))
```

```python
# maximum and minimum :
import numpy as np
a = np.array([2,23,45,76,12,91,101,1100])
x1 = np.max(a)
x2 = np.min(a)
print(x1)
print(x2)
x3 = np.amax(a)
print(x3)
```

```python
# percentile:
a = np.array([20,10,45,1,2,6])
x1 = np.percentile(a,50)
print(x1)
```

```python
# <,> , == , != , <= ,>=
import numpy as np                                        # we are also use less(), less_equal() , greater() ,greater_equal() , equal(),not_equal()
x = np.array([1,2,4,6,8])
y = np.array([3,5,8,1,8])
result = x < y
print(result)
result1 = x > y
print(result1)
result2 = x == y
print(result2)
result3 = x != y
print(result3)
```

```python
# where command:
import numpy as np
a = np.array([23,45,67,89,91,90,20])
result = np.where( a > 60  )
print(result)

result2 = np.where(a % 2 == 0)
print(result2)

result3 = np.where(a > 60 , a*2 , a-2)       # a > 60 toh a * 2 , verna a -2 kar do
print(result3)
```
