```python
# example :
n = numpy.arange(12)
d = n.reshape(3,4)              # 3 -> Rows and 4 -> colunms
print(type(d))
print(d)

# another way :
n = numpy.arange(12).reshape(4,3)
print(n)

# 3d :
h = numpy.arange(24).reshape(2,3,4)
print(h)
```

```python
import numpy
n = numpy.arange(12).reshape(4,3)
print(n)
print(n[0:4,2:4])          # one way
print(n[0:4:,2])           # second way -> by jump
print(n[1:3,1:2])
print(n[1:3,1])
```
