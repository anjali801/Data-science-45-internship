```python
# -5 to 256 ke andar agar number hai ton python same memory per point karta hai
# agar ise range se bahar hai toh same memory per ponit nahi karega

# Example 1: Numbers within the -5 to 256 range (negative)
c = -4
d = -4
print(f"For c={c}, d={d}:")
print(f"c == d: {c == d}")
print(f"c is d: {c is d}") # Should be True due to caching
print(f"c is not d: {c is not d}")
print("---")

# Example 2: Numbers outside the -5 to 256 range (negative)
e = -6
f = -6
print(f"For e={e}, f={f}:")
print(f"e == f: {e == f}")
print(f"e is f: {e is f}") # Should be False outside caching range
print("---")

# Example 3: Numbers within the -5 to 256 range (positive)
g = 100
h = 100
print(f"For g={g}, h={h}:")
print(f"g == h: {g == h}")
print(f"g is h: {g is h}") # Should be True due to caching
print("---")

# Example 4: Numbers outside the -5 to 256 range (positive)
i = 257
j = 257
print(f"For i={i}, j={j}:")
print(f"i == j: {i == j}")
print(f"i is j: {i is j}") # Should be False outside caching range
print("---")
```
