# IN the case of the Rows there will be  : -
(add the rows)

1. loc  : - Label based->  in this you need to pass column name and index

2. iloc : - index based -> in this need to pass the index numbers of the value

```python
import pandas as pd
d = {
    "name":["anna","jiva","chinmay","shiv"],
    "salary":[100,200,300,400]
}
df = pd.DataFrame(data=d)

# getting single element :-
print(df.loc[2,"name"])
print(df.iloc[2,0])

# getting single row :-
print(df.loc[2])

# getting multiple rows using iloc :-
print("\n",df.iloc[0:3])

# getting multiple rows using loc :-
print("\n",df.loc[0:3])
print("\n",df.loc[0:3,"name"])
print("\n",df.loc[0:3,"salary"])
```

```python
#  getting sub data set  :-
print(df)
# by loc :
df1 =df.iloc[0:2,[0]]
print("\n", df1)
# by iloc :
df2 = df.loc[0:2,["salary"]]
print("\n",df2)
```
