```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
     "salary" : [23,45,67,89,12,35,56]
     }
df = pd.DataFrame(data)
df.shape

# Line plot :
plt.plot(df["salary"], linestyle = "--",color = "blue",linewidth = "3")
plt.show()
```

```python
# Histogram :
```

```python
plt.hist(df["salary"], bins = 3 , color="pink")
plt.show()
```

```python
# Boxplot :
```

```python
plt.boxplot(df["salary"])
plt.show()
```

```python
df.drop(index = 6 ,inplace = True)
df
```

```python
# for categorical column or values : -
```

```python
df["dept"] = ["HR","IT","Finance","IT","Finance","HR"]
df
```

```python
# Pie chart:
count = df["dept"].value_counts()
count
```

```python
plt.pie(count)
plt.show()
```

```python
plt.pie(count , labels = count.index)
plt.show()
```

```python
plt.pie(count , labels = count.index , autopct = "%1.1f")
plt.show()
```

```python
plt.pie(count , labels = count.index , autopct = "%1.0f")
plt.show()
```

```python
plt.pie(count , labels = count.index , autopct = "%1.1f" , explode = [0,0.1,0])
plt.show()
```

```python
plt.pie(count , labels = count.index , autopct = "%1.1f" , explode = [0,0.1,0.2])
plt.axis("equal")
plt.show()
```

```python
# Countplot :
plt.bar(count.index , count)
plt.show()
```

```python
plt.bar(count.index , count , color = ["green" , "pink","brown"])
plt.show()
```

# Bivariate Analysis : -    
it basically try to analysis the relation between two feature and column

-> numerical - numerical column :

```python
df["Age"] = [12,23,14,34,45,11]
df
```

```python
# Scatter plot :
plt.scatter(df["salary"] , df["Age"])
plt.grid()
plt.show()
```

```python
# line plot :
plt.plot(df["salary"] , df["Age"])
plt.grid()
plt.show()
```

```python
sort_age = df.sort_values(by = "Age")
sort_age
```

```python
plt.plot(df["salary"],sort_age["Age"])
plt.grid()
plt.show()
```

```python
# bar chart :
plt.bar(df["salary"],sort_age["Age"])
plt.grid()
plt.show()
```

-> Numerical - categorical :

```python
hr_sal = df[df["dept"]=="HR"]["salary"]
hr_sal
```

```python
IT_sal = df[df["dept"]=="IT"]["salary"]
IT_sal
```

```python
Finance_sal = df[df["dept"]=="Finance"]["salary"]
Finance_sal
```

```python
# box plot :
plt.boxplot([hr_sal ,IT_sal ,Finance_sal], labels = ["HR","IT","Finance"])
plt.show()
```

```python
# pie chart :
salary_by_dept = df.groupby("dept")["salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept , labels = salary_by_dept.index , autopct = "%1.1f" )
plt.axis("equal")
plt.show()
```

```python
# bar plot :
hr_mean = sum(hr_sal)/len(hr_sal)
IT_mean = sum(IT_sal)/len(IT_sal)
Finance_mean = sum(Finance_sal)/len(Finance_sal)
plt.bar(["HR","IT","Finance"],[hr_mean,IT_mean,Finance_mean], color = ["red" ,"yellow","purple"])
plt.show()
```
