```python
import numpy as np
# 2d list
l = [[1,2,3],[4,5,6]]
print(l)
print(l[0])       # complete first row
print(l[0][0])    # first element of first row
print(l[0][1])    # second element of the first row
l[1][0] = 100
print(l)

# 2d array
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print(arr[0])      # complete first row
print(arr[0][0])   # first element of first row
arr[1][0] = 100
print(arr)
```

```python
# comparsion
# list
import time
start = time.time()
l = [i*2 for i in range(100000)]
print("list output:",time.time()-start)

# array
import numpy as np
start = time.time()
arr = np.arange(100000)*2
print("array output:",time.time()-start)
```

```python
# zeros array :-
import numpy as np
arr = np.zeros(5)
print( "zeros array 1D",arr)
arr1 = np.zeros((5,2))
print("zeros array 2D","\n",arr1)
```

```python
# ones array :
import numpy as np
arr = np.ones(5)
print( "ones array 1D",arr)
arr1 = np.ones((5,2))
print("ones array 2D","\n",arr1)
```

```python
import numpy as np
arr = np.zeros((3,4))
print(arr)
arrm = arr + 10
print("\n",arrm)


arr1 = np.ones((3,4))
print("\n",arr1)
arrm1 = arr1 * 10
print("\n",arrm1)
```

```python
# full :-
import numpy as np
arr = np.full((3,4),10)         # define row and column and value
print(arr)

arr1 = np.full((4),11)
print("\n",arr1)
```
