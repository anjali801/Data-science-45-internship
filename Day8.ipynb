```python
import numpy as np
# 2d list
l = [[1,2,3],[4,5,6]]
print(l)
print(l[0])       # complete first row
print(l[0][0])    # first element of first row
print(l[0][1])    # second element of the first row
l[1][0] = 100
print(l)

# 2d array
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print(arr[0])      # complete first row
print(arr[0][0])   # first element of first row
arr[1][0] = 100
print(arr)
```

```python
# comparsion
# list
import time
start = time.time()
l = [i*2 for i in range(100000)]
print("list output:",time.time()-start)

# array
import numpy as np
start = time.time()
arr = np.arange(100000)*2
print("array output:",time.time()-start)
```

```python
# ones array :
import numpy as np
arr = np.ones(5)
print( "ones array 1D",arr)
arr1 = np.ones((5,2))
print("ones array 2D","\n",arr1)
```

```python
# vector (1d)
# matrix (2d)
# tensor (3d)

import numpy as np

print("Vector:-")
# vector 1d list :
l = [1,2,3]
print(l)
# vector 1d array :
arr = np.array([2,4,6,8])
print("\n",arr)


print("\n","Matrix :-")
# matrix 2d list :
l =[[1,2,3],[4,5,6]]
print(l)
# matrix 2d array :
arr2 = np.array([[1,2,3],[4,5,6]])
print("\n",arr2)

print("\n","Tensor:-")
# tensor 3d list :
l = [[[1,2],[3,4]],[[5,6],[7,8]]]
print(l)
# tensor 3d array :
arr3 = np.array(l)
print("\n",arr3)
```

```python
# Array Properties : -
import numpy as np
arr = np.arange(11)                    # 1d array
print("shape :",np.shape(arr))         # one way
print("shape :",arr.shape)             # another way
print("size :",np.size(arr))
print("dimension :",np.ndim(arr))
print("data type :",arr.dtype)

print("\n","Another example:-")
arr = np.arange(12).reshape(3,4)      # 2d array
print("shape :",np.shape(arr))
print("size :",np.size(arr))
print("dimension :",np.ndim(arr))
print("data type :",arr.dtype)
```
