# Multivariate Analysis :        
where we try to understand how three columns are affect and change to each other

```python
# numerical - numerical - numerical
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
     "salary" : [23,45,67,89,12,35,56] ,
     "Age" : [30,23,45,34,23,21,31] ,
     "exp" : [12,5,20,17,7,10,12],
     "dept": ["HR","IT","HR","IT","HR","IT","HR"]
     }
df = pd.DataFrame(data)
df
```

```python
# Bubble plot :
plt.scatter(df["Age"],df["salary"],s=df["exp"] , color = "skyblue" , edgecolors = "black")
plt.show()
```

```python
# 2 numerical and 1 categorical column
```

```python

plt.scatter(df["Age"], df["salary"], c=df["dept"].map({"HR": "yellow", "IT": "red"}))
plt.xlabel("Age")
plt.ylabel("salary")
plt.title("Age vs Salary  vs Dept")
plt.legend
plt.show()
```

```python
#questio multiple line chart


import matplotlib.pyplot as plt

x = [2000, 2005, 2010, 2015]
y1 = [110, 140, 150, 300]
y2 = [100, 321, 150, 653]
y3 = [200,400,100,50]

plt.plot(x, y1,
         label="jeans",
         color="green",
         marker="o",
         linestyle="--",
         linewidth=4,
         markersize=10)

plt.plot(x, y2,
         label="T-shirt",
         color="pink",
         marker="*",
         linestyle="-",
         linewidth=4,
         markersize=10)

plt.plot(x, y3,
         label="shirt",
         color="red",
         marker="o",
         linestyle=":",
         linewidth=2,
         markersize=10)


plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

groups = ["group A", "group B", "group C", "group D"]
y1 = [10, 20, 20, 40]
y2 = [20, 30, 25, 30]
y3 = [23, 35, 37, 40]
y4 = [10, 20, 20, 40]
# y5 = [20, 30, 25, 30]
# y6 = [23, 35, 37, 40]

x = np.arange(len(groups))
width = 0.30

plt.bar(x - width, y1, width, label="boys", color="red")
plt.bar(x , y2, width, label="boys", color="black")
plt.bar(x + width, y3, width, label="girls", color="blue")
plt.bar(x + width, y4, width, label="girls", color="green")

plt.xlabel("groups")
plt.ylabel("number of students")
plt.title("No. of students in each group")
plt.xticks(x, groups)
plt.legend()

plt.show()
```

# Oject Oriented API :           
in this we just create the object before starting the ploting , basically allows subplots , multiple axis & multiple plots in the one - figure .


```python
fig ,axs =plt.subplots(1,3,figsize =(15,5))
```

```python
axs[0].plot(df["Age"],df["salary"])
axs[0].grid()
axs[0].set_title("Line plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("salary")

axs[1].scatter(df["Age"],df["salary"])
axs[1].grid()
axs[1].set_title("Scatter plot")
axs[1].set_xlabel("Age")
axs[1].set_ylabel("salary")

axs[2].bar(df["Age"],df["salary"])
axs[2].set_title("Bar plot")
axs[2].set_xlabel("Age")
axs[2].set_ylabel("salary")

plt.show()

```

```python
# Another Example :
data = {
    "year" : [2023 , 2020, 2024 ,2025,2021],
    "sales": [100,200,300,400,500],
    "profit": [23,45,67,12,67],
    "expense":[34,56,78,90,45]
}

df = pd.DataFrame(data)
df
```

```python
plt.plot(df["year"],df["sales"],label = "sales")
plt.plot(df["year"],df["profit"], label = "profit")
plt.plot(df["year"],df["expense"],label = "expense")
plt.title("Financial Analysis")
plt.xlabel("year")
plt.ylabel("amount")
plt.legend()
plt.show()
```

```python
plt.bar(df["year"],df["sales"],label = "sales")
plt.bar(df["year"],df["profit"], label = "profit")
plt.bar(df["year"],df["expense"],label = "expense")
plt.title("Financial Analysis")
plt.xlabel("year")
plt.ylabel("amount")
plt.legend()
plt.show()
```
