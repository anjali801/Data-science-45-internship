```python
# subplot ::-->

import matplotlib.pyplot as plt
import numpy as np

# first chart:
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.subplot(1,2,1)      # first - rows , second - column , third - position
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')



# second chart:
x1 = [1,2,3,4,5]
y1 = [100,200,300,400,500]
plt.subplot(1,2,2)
plt.pie(y1,labels = x1)
plt.xlabel('x1-axis')
plt.ylabel('y2-axis')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# first chart:
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.subplot(2,2,1)      # first - rows , second - column , third - position
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')



# second chart:
x1 = [1,2,3,4,5]
y1 = [100,200,300,400,500]
plt.subplot(2,2,2)
plt.pie(y1,labels = x1)
plt.xlabel('x1-axis')
plt.ylabel('y2-axis')


# Third chart:
plt.subplot(2,2,3)
plt.bar(x,y)


#fouth chart:
plt.subplot(2,2,4)
plt.scatter(x,y)
plt.tight_layout()
plt.show()
```
