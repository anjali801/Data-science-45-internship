Mini Project1 :Student Data Manager


```python
class student :
    def __init__(self ,Stu_ID, Name , Age ,course ):
        self.Stu_ID = Stu_ID
        self.Name = Name
        self.Age = Age
        self.course = course

    def display(self):
        print("Student ID :" , self.Stu_ID)
        print("Name :" , self.Name)
        print("Age:",self.Age)
        print("course:",self.course)

students = []

def add():
        StuId = input("Enter the student ID:")
        Name = input("Enter the  Name :")
        Age = input("Enter the age :")
        course = input("Enter the course :")

        students.append(student(StuId,Name,Age,course))
        print("Student added successfully")

def delete():
  sid = input("Enter the student Id:")
  for i in students:
            if i.id == sid:
                students.remove(i)
                print("Student deleted successfully")
            else:
              print ("Student not found")



def search():
  sid = input("Enter the student Id:")
  for i in students:
            if i.id == sid:
                i.display()
            else:
              print("Student not found")


def update():
  sid = input("Enter the student Id:")
  for i in students:
        if i.id == sid:
            i.name = input("Enter the name :")
            i.age = input("Enter the age :")
            i.course = input("Enter the course :")
            print("Student updated successfully")
        else:
          print("Student not found")

while True:
 print("1.Add Student")
 print("2.Delete Student")
 print("3.Search Student")
 print("4.Update Student")
 print("5.Exit")
 choice = input("Enter the choice :")
 if choice == "1":
    add()
 elif choice == "2":
    delete()
 elif choice == "3":
    search()
 elif choice == "4":
    update()
 elif choice == "5":
    break
 else:
    print("Invalid choice")







```
