```python
# reshape:
import numpy as np
arr = np.arange(24)
up_arr = np.reshape(arr, (3 , 8))
print(up_arr)

arr1 = up_arr.flatten()             # flatter convert 1d & 2d array into 1d array
print(arr1)                          # create copy then work
print(up_arr)
```

```python
# Example 3d :
arr = np.arange(24)
up_arr = np.reshape(arr, (2 , 3 , 4))
print(up_arr)
arr1 = up_arr.flatten()
print("\n",arr1)
print("\n",up_arr)
```

```python
# ravel :                            # ravel is like a flatter but there are some small defference  is flatter work on copy and ravel work on original means effect the original value
import numpy as np                    # work on original array
arr = np.arange(14).reshape(7,2)
print(arr)
arr1 = np.ravel(arr)
print("\n",arr1)
print("\n",arr)
```

```python
# for 3d :
import numpy as np
arr = np.arange(24).reshape(2,3,4)
print(arr)
arr1 = np.ravel(arr)
print("\n",arr1)
```

```python
# transpose :
import numpy as np                                # in this method rows convert into columns
arr = np.arange(12).reshape(3,4)                   # columns convert int rows
print(arr)
arr1 = np.transpose(arr)
print("\n",arr1)
```

```python
# slicing  :-

# 1D :
import numpy as np
arr = np.arange(11)
print(arr)
up_arr = arr[1:4]
print("\n",up_arr)

# 2D :
arr1 = np.arange(16).reshape(8,2)
print("\n",arr1)
up_arr1 = arr1[1:4,1:3]
print("\n",up_arr1)

# 3D :
arr2 = np.arange(24).reshape(2,3,4)
print("\n",arr2)
up_arr2 = arr2[1:2,1:3,1:3]
up1 = arr2[0:1,0:1,0:1]
print("\n",up_arr2)
print("\n",up1)


```
