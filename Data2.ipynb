```python
# function:- keyword used to define a function is def
# receiver is a parameter
# argument is whwn you pass the value

def hello():
  print("hello function is working")


hello()
```

```python
def hello1(a): # here is parameter
  print(a)

hello1(100)  # here is argument



# Another example
def add(a=0,b=0):
  print(a+b)
add(10,30)
add()

# Example
def power(a,b):
  print(a**b)

power(5,2)
power(2,5)
power(a=5,b=2)
power(a=2,b=5)

# Another Example
def student(*a):                          # *a used to get multiple parameter
  print(a)
  print(type(a))
student(1,3,5,7,9)

# In python if have same name function the call the function it's only call the last function



```

```python
def marks(*a):
  for i in range(len(a)):
    print(a[i])

marks(1,2,3,4,5,6)

print(" For the list :")
def mark(a):
  for i in a:
    print(i)

mark([1,2,3,4,5,6])

print("\nAnother Example :")
def address(z):
  for i in range(len(z)):
    print("\n")
    for j in range(len(z[i])):
     print(z[i][j])

address(["hello","Anjali"])


print("\nAnother:")
def sum(a,b):
  return a+b

result = sum(10,20)
print(result)
```

```python
# Lambda function :-[lambda arguments : expression]
add = lambda x: x
print(add(0))
sum = lambda x,y: x+y
print(sum(10,20))

a = lambda x:x
print(a([10,20,30,40]))

b = lambda *x:x
print(b(10,34,45,78,98))
```

```python
# List comprehension [expression for item in iterable]
print([i for i in range(5)])

print("\nAnother Example :")
l = [10,20,30,40]
print([l[i] for i in range(len(l))])

print("\n Another one :")
k = [10,20,[30,40,50]]
print([k[2][i]  for i in range(len(k[2]))])
```

```python
# list
l = [1000,98,39,340 , 200, 250 , 200]
print("list is :" , l)
print("length of the list :" , len(l))
l.insert(2,500)
print("After add a value at particular index:",l)
l.append(50)
print("After the Adding the value in the last :",l)
l.reverse()
print("Reverse of list :",l)
l.sort()
print("ascending order:",l)
l.sort(reverse = True)
print("Decending order:",l)
l.remove(200)
print("after removing : " , l)
l.pop(1)
print("After removing by giving index number:",l)

```
