```python
# oops concepts
# defination
# Example :-

class anjali:
  def __init__(self,name):                            # __init__ is a constructor   # here name is instance variable
    self.name = name                                  # self is used to access the value all over the class
  def show(self):
    print(self.name)

r = anjali("Anna")
r.show()
```

```python
# Another Example :-

class anna:
  def __init__(self):
    print("calling cnstructor")

  def show(self):
    print("show the name")

r = anna()
r.show()


# Another Example :-

class anna:
  def __init__(self , name ,age):
    self.name = name
    self.age = age

  def getage(self):
    print("My age is :",self.age)

  def getname(self):
    print("My name is :",self.name)

# r = anna("Anna",20)
# r.getage()
# r.getname()

r = anna(age=30,name="jiva")
r.getage()
r.getname()
```

```python
class student:
  def __init__(self,*args):
    self.data = args
  def user(self):
    return self.data[0]                               # direct yaha per bhi loop laga sakhte the


  def detail(self):
    return self.data[1]                             # yaha bhi  loop laga sakhate hai


s = student(["dhreeraj","kunal","harsh","praveen"],{"address":"kukas","college":"Arya", "location":"jaipur"})
t = s.user()
x = s.detail()

for i in t:
  print(i)

print("address is :", x["address"])
print("college is :", x["college"])
print("location is:", x["location"])


```
