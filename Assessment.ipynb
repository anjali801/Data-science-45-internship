```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url="https://raw.githubusercontent.com/anjali801/Data-science-45-internship/main/data.json"
df = pd.read_json(url)
print(df)
```

```python
df["revenue"] = df["price"] * df["quantity"]
df
```

```python
print("Average Price")

print(np.mean(df["price"]))

print("\n", "Maximum Revenue")

print(np.max(df["revenue"]))

print("\n","Minimum Revenue")

print(np.min(df["revenue"]))
```

```python
category1 = df.groupby(
    "category"
)["revenue"].sum()
print(category1)
```
