#K-Means Clustering :

-> most popular clustering algorithm

-> It divides data into K clusters

-> k --> no of clusters

#note:-
 Elbow method -> helps to find the best value of k.

```python
# clustering
import pandas as pd
from sklearn.cluster import KMeans

data = {
    "age": [22,25,24,45,50,48,20,47],
    "income": [22000,25000,24000,45000,50000,48000,20000,47000]
}

df = pd.DataFrame(data)
print(df)

# input| feature
x = df[['age','income']]

# model
model = KMeans(n_clusters=2,random_state=42)
model.fit(x)

# clusters
clusters = model.labels_
print(clusters)
df['clusters'] = clusters
print()
```

```python
# example
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

data = {
    "age": [22,25,24,45,50,48,20,47],
    "income": [22000,25000,24000,45000,50000,48000,20000,47000]
}

df = pd.DataFrame(data)
print(df)

# input| feature
x = df[['age','income']]

# model
model = KMeans(n_clusters=2,random_state=42)
model.fit(x)

# clusters
df['clusters'] = model.labels_

plt.scatter(df['age'],df['income'],c=df['clusters'])
plt.xlabel('age')
plt.ylabel('income')
plt.title('Group the Data')
plt.show()
```

```python
# example
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Load dataset
iris = load_iris()

X = iris.data

# KMeans
kmeans = KMeans(n_clusters=3, random_state=42)

clusters = kmeans.fit_predict(X)

# Plot
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=clusters,
    cmap="viridis"
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means on Iris Dataset")

plt.show()
```
