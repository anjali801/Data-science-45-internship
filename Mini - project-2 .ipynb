```python
# drop row 3 -> day = wednesday
import pandas as pd
url = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-arya/main/temperature_data.json"
df = pd.read_json(url)
df
df.drop(2 , inplace = True)
df
```

```python
# fill average value -> humidity -> nan
# sum1 = sum(df['humidity_pct'])
# avg = sum1/len(df['humidity_pct'])
# print(avg)
# df['humidity_pct'].fillna(avg, inplace = True)
# df
df["humidity_pct"]= df["humidity_pct"].fillna(df["humidity_pct"].mean())
df
```

```python
# new cols -> fareheit  | cell*l.8 -> +32
df['fareheit'] = df['temperature_c']*1.8+32
df
```

```python
# subplots -> pie
import matplotlib.pyplot as plt
import numpy as np
fig, axs = plt.subplots(1, 2, figsize=(15, 5))

axs[0].pie(df["temperature_c"], autopct = "%1.1f%%" , labels = df["day"] , shadow = True )
axs[0].set_title("Temperature")

axs[1].pie(df["humidity_pct"], autopct = "%1.1f%%" ,labels=df["day"])
axs[1].set_title("Humidity")

fig.savefig("pie.png")

plt.show()
```
