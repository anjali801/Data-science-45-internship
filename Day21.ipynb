# Line plot :               

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

var = [1,2,3,4,5,6,7]
var_1 = [8,9,10,11,12,13,14]

x_1 = pd.DataFrame({"var" : var , "var_1":var_1})

sns.lineplot(x="var",y="var_1",data=x_1)             # here data ----uses because  it save or take the dataframe
plt.show()
```

```python
plt.plot(var,var_1)        # matplotlib graphs
plt.show()
```

```python
var = [1,2,3,4,5,6,7]
var_1 = [8,9,10,11,12,13,14]

sns.lineplot(x=var,y=var_1)
plt.show()
```

```python
p_1 = sns.load_dataset('penguins').head(25)
p_1
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1)
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1,hue="sex")
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1,hue="sex" , size=10)
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1,hue="sex", style="sex")
plt.show()
```

```python
# palette - through this change the colour
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1,hue="sex",style="sex", palette="Accent")
plt.show()
```

```python

sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1,hue="sex",style="sex", markers=["o",">"])
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1,hue="sex",style="sex", markers=["o",">"],dashes=False)
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1,hue="sex",style="sex", markers=["o",">"],legend=False)
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=p_1,hue="sex",style="sex", markers=["o",">"])
plt.grid()

plt.show()
```

# Histogram plot :
-> distplot()   used .......

```python
sns.distplot(p_1["bill_length_mm"])
plt.show()
```

```python
sns.distplot(p_1["flipper_length_mm"])
plt.show()
```

```python
sns.distplot(p_1["flipper_length_mm"], bins=[170,180,190,200,210,220,230,240])
plt.show()
```

```python
sns.distplot(p_1["flipper_length_mm"],kde=False)                  # kde -> line show for true by default
plt.show()
```

-> log_scale parameter used to represent graph at log scale .............

```python
sns.distplot(p_1["flipper_length_mm"] , kde =True , rug=True)
plt.show()
```

```python
sns.distplot(p_1["flipper_length_mm"],color="red")
plt.show()
```
