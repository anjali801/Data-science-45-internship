# Encoding :
categariol data   convert  into numerical data .

--> one hot encoding :------------------------------ use : when short number of the data is present  (ex - male & female , Yes & No)

```python
import pandas as pd
data = {
    "name" : ["A","B","C","D","E","F","G"],
    "gender" : ["male","female","male","female","male","female","male"]
}
df = pd.DataFrame(data)
print(df)
```

```python
df.isnull().sum()
```

# : encoding by get dummies

```python
pd.get_dummies(df['gender'])            # check by info then shows  boolean data
```

#: encoding by OneHotEncoder class

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder()
ohe_e = ohe.fit_transform(df[['gender']])         # give sparse matrix
print(ohe_e)
```

```python
ohe_e1 = ohe_e.toarray()
print(ohe_e1)
```

```python
pd.DataFrame(ohe_e1,columns = ohe.categories_)
```

# Sparse Matrix  :  is the matrix in which most of the element is 0 & 1

```python
#handle missing values :
import pandas as pd
import numpy as np

data = {
    "colors" : ['red','green','blue','orange','green','blue',np.nan]
}
df = pd.DataFrame(data)
print(df)


#handle null values :
df.dropna(inplace = True)
print(df)
```

```python
# label Encoding :
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['colors_encoder'] = le.fit_transform(df['colors'])
print(df)
```

```python
# one hot encoding :
from skelarn.preprocessing import OneHotEncoder
print(df)

# drop cols of label encoding
if 'colors_encoder' in df.keys():
  df.drop(columns = ['colors_encoder'],inplace = True)
  print(df)
```

```python
# one host encoding :(by the help of sklearn) :
df_encoder  = pd.get_dummies(df['colors'])
print(df_encoder)
```

```python
# combination of handle missing values |label encoding  and  onehostEncoding
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder

data = {
    "age":[12,13,14,15,16,17,np,nan],
    "gender":["male","female","male","female","male","female",None],
    "name": ['A','B','C','D','E','F','G'],

}
df = pd.DataFrame(data)

# handle missing values
df['age'] = df['age'].fillna(df['age'].mean())
df.dropna(subset = ['gender'],inplace= True)
print(df)

# label encoding
le = LabelEncoder()
df['gender_1'] = le.fit_transform(df['gender'])
print(df)

# one hot encoding
oe = OneHotEncoder()
oe_e = oe.fit_transform(df[['gender']])
df_oe = pd.DataFrame(oe_e.toarray(),columns = le.classes_)
print(df_oe)
```
